## Required Components <a class="anchor" id="components"></a>

To run this demonstration, you will require the following equipment included with the ZCU208 development kit:

* [RFSoC ZCU208 development board](#zcu208-annotated).
* [XM655 add-on card](#zcu208-annotated).
* [CLK104 add-on card](#zcu208-annotated). 
* At least 2 [Carlisle cables](#carlisle).

### Minimum Setup (2 Carlisle cables)

If you only have two Carlisle cables, you can synchronise **4 RF-DAC channels** and **4 RF-ADC channels**. The following additional equipment is required:

* At least **8 high-quality [male-to-female DC blockers](#DCBLK-SMAF)** of the same model.  
* At least **8 high-quality [female-to-female SMA adapters](#DCBLK-SMAF)** of the same model.

### Full Setup (4 Carlisle cables)

To synchronise all **8 RF-DAC channels** and **8 RF-ADC channels**, you will need **4 Carlisle cables** and the following additional equipment:

* **16 high-quality [male-to-female DC blockers](#DCBLK-SMAF)** of the same model.  
* **16 high-quality [female-to-female SMA adapters](#DCBLK-SMAF)** of the same model.

## Setup <a class="anchor" id="setup"></a>
Unlike other boards, the ZCU208 requires the CLK104 and XM655 add-on cards to configure the clocks and RFDCs, respectively. This section describes how to connect the CLK104 clocks to the development board and how to access the RFDCs via SMA cables from the XM655 board. Using the previously mentioned DC blockers and SMA adapters, the board will be configured in loopback mode.

### CLK104 Setup <a class="anchor" id="clk104"></a>
On the CLK104 card, make the connections shown in [Figure 4](#figure-4), such that

* DAC_229_CLK is connected to DAC_RFCLK_A (P & N).
* ADC_225_CLK is connected to ADC_RFCLK_A (P & N).

Incorrect connections will cause the demonstration to fail.

### Loopback Setup <a class="anchor" id="loopback"></a>

Next, connect the RF-DAC SMA outputs to the RF-ADC SMA inputs to create a loopback path.

For this demonstration, the on-board baluns on the XM655 should **not** be used. Each RF-DAC/RF-ADC pair shares analogue filtering through a limited number of baluns, which can impact synchronisation. Instead, we directly connect the differential signals exposed via the Carlisle cables.

#### Differential Connection Rules

To correctly form a loopback:

* Connect DAC tile (e.g. 228–231) **P &rarr;** ADC tile (e.g. 224–227) **P**
* Connect DAC tile (e.g. 228–231) **N &rarr;** ADC tile (e.g. 224–227) **N**

Both signals from a given DAC tile must be connected to the **same ADC tile**.  
Splitting a differential pair across multiple ADC tiles will result in incorrect operation.

#### Example

If connecting RF-DAC tile 230 to RF-ADC tile 224:

* DAC tile 230 **P** &rarr; ADC tile 224 **P**  
* DAC tile 230 **N** &rarr; ADC tile 224 **N**

#### Tile Mapping

For this demonstration, **any DAC tile can be connected to any ADC tile**, provided differential pairs are preserved. A placeholder mapping is shown below:

**Using 2 Carlisle cables (JHC1 & JHC5 only):**
* DAC tile 228_0 &rarr; ADC tile 224_0
* DAC tile 228_2 &rarr; ADC tile 224_2
* DAC tile 229_0 &rarr; ADC tile 225_0
* DAC tile 229_2 &rarr; ADC tile 225_2

**Using 4 Carlisle cables (JHC1, JHC3, JHC5 & JHC7):**
* DAC tile 228_0 &rarr; ADC tile 224_0
* DAC tile 228_2 &rarr; ADC tile 224_2
* DAC tile 229_0 &rarr; ADC tile 225_0
* DAC tile 229_2 &rarr; ADC tile 225_2
* DAC tile 230_0 &rarr; ADC tile 226_0
* DAC tile 230_2 &rarr; ADC tile 226_2
* DAC tile 231_0 &rarr; ADC tile 227_0
* DAC tile 231_2 &rarr; ADC tile 227_2

#### Physical Connections

For each differential P/N pair:

- Insert a **male-to-female DC blocker**
- Connect this to a **female-to-female SMA adapter**

This allows the Carlisle cable SMA connectors (male) to be safely bridged from RF-DAC outputs to RF-ADC inputs while blocking DC components.
