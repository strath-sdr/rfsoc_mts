## Required Components <a class="anchor" id="components"></a>

To run this demonstration, you will require:

* [RFSoC 4x2 development board](#rfsoc4x2)
* 2 high quality [**length-matched** male-to-male SMA cables](#sma)

### Standard Setup (2-channel demonstration)

This notebook demonstrates synchronisation between RFDC channels located on different tiles. Since channels on the same RFDC tile are inherently aligned without requiring the Multi-Tile Synchronisation (MTS) engine, only two RF-DAC channels and two RF-ADC channels are required.

### Extended Setup (4-channel ADC observation)

If you wish to observe synchronisation across all four RF-ADC channels simultaneously, additional hardware is required:

* A total of 6 high-quality **length-matched** male-to-male SMA cables
* 2 × 2-way RF splitters

## Setup <a class="anchor" id="setup"></a>

This section describes how to connect the RFSoC 4x2 development board in loopback mode for the MTS demonstration.

The objective is to evaluate synchronisation between signals transmitted from RF-DAC channels on different DAC tiles and received by RF-ADC channels on different ADC tiles.

### RFDC Channel Mapping <a class="anchor" id="mapping"></a>

The RFDC channels are mapped to the RFSoC 4x2 board as follows:

* ADC tile 224_0 &rarr; ADC_D
* ADC tile 224_2 &rarr; ADC_C
* ADC tile 226_0 &rarr; ADC_B
* ADC tile 226_2 &rarr; ADC_A

* DAC tile 230_0 &rarr; DAC_A
* DAC tile 228_0 &rarr; DAC_B

### Loopback Setup <a class="anchor" id="loopback"></a>
#### Connection Rules

Connect each of the two RF-DAC outputs to an RF-ADC input using **length-matched SMA cables**. For the clearest demonstration of MTS behaviour, the selected RF-ADC channels should reside on different ADC tiles, as channels within the same tile are natively synchronised. If two RF-ADC channels from the same tile are used, the setup can still demonstrate synchronisation between DAC_A and DAC_B, since these reside on different DAC tiles; however, this does not demonstrate synchronisation behaviour across ADC tiles.

Example:

* DAC_A &rarr; ADC_D (tile 224)
* DAC_B &rarr; ADC_B (tile 226)
